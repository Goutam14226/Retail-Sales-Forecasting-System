```python
DATA_PATH = os.path.join(
    PROJECT_ROOT,
    "data",
    "features_production_clean.parquet"
)

MODEL_PATH = os.path.join(
    PROJECT_ROOT,
    "models",
    "lstm_attention_production.keras"
)

SCALER_PATH = os.path.join(
    PROJECT_ROOT,
    "models",
    "feature_scaler.pkl"
)

FEATURE_COLUMNS = [
    "sales",
    "sell_price",
    "price_available",
    "day_of_month",
    "week_of_year",
    "day_of_year",
    "quarter",
    "is_weekend",
    "lag_1",
    "lag_7",
    "lag_14",
    "lag_28",
    "rolling_mean_7",
    "rolling_mean_28",
    "rolling_std_7",
    "rolling_std_28",
    "is_event_day",
    "event_count",
    "snap_active"
]

engine = ForecastEngine(
    data_path=DATA_PATH,
    model_path=MODEL_PATH,
    scaler_path=SCALER_PATH,
    feature_columns=FEATURE_COLUMNS,
    sequence_length=28
)

print("ForecastEngine initialized successfully")
print("Model:", engine.model.name)
print("Parameters:", engine.model.count_params())
print("Features:", len(engine.feature_columns))
print("Sequence length:", engine.sequence_length)
print("Row groups:", engine.parquet_file.num_row_groups)

print("FORECAST ENGINE INITIALIZATION TEST: PASS")
```

```python
from fastapi import FastAPI, HTTPException
from pydantic import BaseModel
import threading
import uvicorn


class ForecastRequest(BaseModel):
    item_id: str
    store_id: str


app_v2 = FastAPI(
    title="M5 Sales Forecasting API",
    description="API for next-day sales forecasting using LSTM with Attention",
    version="1.0.0"
)


@app_v2.get("/")
def root():
    return {
        "message": "M5 Sales Forecasting API is running",
        "status": "ok"
    }


@app_v2.post("/forecast")
def forecast(request: ForecastRequest):

    try:

        result = engine.forecast_next_day(
            item_id=request.item_id,
            store_id=request.store_id
        )

        return result

    except ValueError as e:

        raise HTTPException(
            status_code=404,
            detail=str(e)
        )


def run_api_v2():
    uvicorn.run(
        app_v2,
        host="0.0.0.0",
        port=8001
    )


api_thread_v2 = threading.Thread(
    target=run_api_v2,
    daemon=True
)

api_thread_v2.start()

print("New API server started on port 8001")
print("API ERROR HANDLING SETUP: PASS")
```

##api.py create

```python
API_FILE = os.path.join(
    API_DIR,
    "api.py"
)

api_code = '''
import os
import sys

from fastapi import FastAPI, HTTPException
from pydantic import BaseModel, Field


# --------------------------------------------------
# Project paths
# --------------------------------------------------

PROJECT_ROOT = os.path.abspath(
    os.path.join(
        os.path.dirname(__file__),
        "..",
        ".."
    )
)

SRC_PATH = os.path.join(
    PROJECT_ROOT,
    "src"
)

if SRC_PATH not in sys.path:
    sys.path.insert(0, SRC_PATH)


# --------------------------------------------------
# Production forecasting engine
# --------------------------------------------------

from forecasting.forecast import ForecastEngine


# --------------------------------------------------
# Production paths
# --------------------------------------------------

DATA_PATH = os.path.join(
    PROJECT_ROOT,
    "data",
    "features_production_clean.parquet"
)

MODEL_PATH = os.path.join(
    PROJECT_ROOT,
    "models",
    "lstm_attention_production.keras"
)

SCALER_PATH = os.path.join(
    PROJECT_ROOT,
    "models",
    "feature_scaler.pkl"
)


# --------------------------------------------------
# Feature configuration
# --------------------------------------------------

FEATURE_COLUMNS = [
    "sales",
    "sell_price",
    "price_available",
    "day_of_month",
    "week_of_year",
    "day_of_year",
    "quarter",
    "is_weekend",
    "lag_1",
    "lag_7",
    "lag_14",
    "lag_28",
    "rolling_mean_7",
    "rolling_mean_28",
    "rolling_std_7",
    "rolling_std_28",
    "is_event_day",
    "event_count",
    "snap_active"
]


# --------------------------------------------------
# Forecast engine
# --------------------------------------------------

engine = ForecastEngine(
    data_path=DATA_PATH,
    model_path=MODEL_PATH,
    scaler_path=SCALER_PATH,
    feature_columns=FEATURE_COLUMNS,
    sequence_length=28
)


# --------------------------------------------------
# FastAPI application
# --------------------------------------------------

app = FastAPI(
    title="M5 Sales Forecasting API",
    description=(
        "API for next-day sales forecasting "
        "using LSTM with Attention."
    ),
    version="1.0.0"
)


# --------------------------------------------------
# Request model
# --------------------------------------------------

class ForecastRequest(BaseModel):

    item_id: str = Field(
        ...,
        min_length=1,
        description="Product/item identifier"
    )

    store_id: str = Field(
        ...,
        min_length=1,
        description="Store identifier"
    )


# --------------------------------------------------
# Root endpoint
# --------------------------------------------------

@app.get("/")
def root():

    return {
        "message": "M5 Sales Forecasting API is running",
        "status": "ok"
    }


# --------------------------------------------------
# Forecast endpoint
# --------------------------------------------------

@app.post("/forecast")
def forecast(request: ForecastRequest):

    try:

        result = engine.forecast_next_day(
            item_id=request.item_id,
            store_id=request.store_id
        )

        return result

    except ValueError as e:

        raise HTTPException(
            status_code=404,
            detail=str(e)
        )
'''

with open(API_FILE, "w") as f:
    f.write(api_code)

print("API file created:")
print(API_FILE)

print("File exists:", os.path.exists(API_FILE))
print("File size:", os.path.getsize(API_FILE), "bytes")

print("PRODUCTION API FILE CREATION: PASS")
```

```python
import requests

BASE_URL = "http://127.0.0.1:8003"

tests = []

# 1. Health check
response = requests.get(f"{BASE_URL}/")
tests.append({
    "test": "Health check",
    "expected": 200,
    "actual": response.status_code,
    "passed": response.status_code == 200
})

# 2. Valid forecast
response = requests.post(
    f"{BASE_URL}/forecast",
    json={
        "item_id": "HOBBIES_1_001",
        "store_id": "CA_1"
    }
)

valid_pass = (
    response.status_code == 200
    and "forecast" in response.json()
    and "forecast_date" in response.json()
)

tests.append({
    "test": "Valid forecast",
    "expected": 200,
    "actual": response.status_code,
    "passed": valid_pass
})

# 3. Invalid series
response = requests.post(
    f"{BASE_URL}/forecast",
    json={
        "item_id": "INVALID_ITEM",
        "store_id": "INVALID_STORE"
    }
)

tests.append({
    "test": "Invalid series",
    "expected": 404,
    "actual": response.status_code,
    "passed": response.status_code == 404
})

# 4. Missing field
response = requests.post(
    f"{BASE_URL}/forecast",
    json={
        "item_id": "HOBBIES_1_001"
    }
)

tests.append({
    "test": "Missing field",
    "expected": 422,
    "actual": response.status_code,
    "passed": response.status_code == 422
})

# 5. Empty strings
response = requests.post(
    f"{BASE_URL}/forecast",
    json={
        "item_id": "",
        "store_id": ""
    }
)

tests.append({
    "test": "Empty strings",
    "expected": 422,
    "actual": response.status_code,
    "passed": response.status_code == 422
})

# 6. Wrong data type
response = requests.post(
    f"{BASE_URL}/forecast",
    json={
        "item_id": 12345,
        "store_id": 123
    }
)

tests.append({
    "test": "Wrong data type",
    "expected": 422,
    "actual": response.status_code,
    "passed": response.status_code == 422
})

print("=" * 60)
print("FINAL API TEST RESULTS")
print("=" * 60)

for test in tests:
    status = "PASS" if test["passed"] else "FAIL"
    print(
        f"{status:5} | "
        f"{test['test']:20} | "
        f"Expected: {test['expected']} | "
        f"Actual: {test['actual']}"
    )

print("=" * 60)

all_passed = all(test["passed"] for test in tests)

print("FINAL API VERIFICATION:", "PASS" if all_passed else "FAIL")
```
