```python
from pathlib import Path

PROJECT_ROOT = Path(
    "/content/gdrive/MyDrive/multivariate-time-series-forecasting"
)

dockerignore_path = PROJECT_ROOT / ".dockerignore"
content = dockerignore_path.read_text(encoding="utf-8")

required_entries = [
    "__pycache__/",
    "*.ipynb",
    ".git/",
    "*.csv",
    "data/sales_train_evaluation.csv",
    "data/sales_train_validation.csv",
    "data/sell_prices.csv",
    "data/features_time.parquet",
    "data/features_lag.parquet",
    "data/features_rolling.parquet",
    "data/features_price.parquet",
    "data/features_event_snap.parquet",
    "logs/",
    "tmp/",
]

missing = [
    entry for entry in required_entries
    if entry not in content
]

# Important production files must NOT be ignored
protected_files = [
    "data/latest_history.parquet",
    "models/lstm_attention_production.keras",
    "models/feature_scaler.pkl",
    "src/api/api.py",
    "src/forecasting/forecast.py",
    "src/frontend/app.py",
]

protected_ignored = [
    entry for entry in protected_files
    if entry in content
]

if not missing and not protected_ignored:
    print("All required ignore rules are present.")
    print("Required production files are not explicitly ignored.")
    print("DOCKERIGNORE CHECK: PASS")
else:
    if missing:
        print("Missing ignore rules:")
        for entry in missing:
            print("-", entry)

    if protected_ignored:
        print("\nProtected files incorrectly ignored:")
        for entry in protected_ignored:
            print("-", entry)

    print("DOCKERIGNORE CHECK: FAIL")
```
